**Goal statement:**

This study calculates the product carbon footprint of a typical Level 2 AC EV charger (wallbox) in accordance with ISO 14067, with reference to PEP ecopassport PSR-0018. Its goals are to identify emission hotspots for reduction, benchmark the results against published declarations of similar products, and design a CDR plan to neutralize the residual emissions. 

The intended audience is readers of this portfolio, including interviewers; if this were a real manufacturer, the audience would be ESG, R&D, and supply chain managers and engineers responsible for emission reduction and green product design. 

Because the study relies on public secondary data and has not undergone third-party critical review, the results support a reasonableness check against other products but not comparative assertions.


## 2. Functional Unit and Reference Flow

**Guiding questions**
- What *function* does an AC Level 2 charger provide to its user?
- Should the unit be "one charger" or something tied to the function delivered? What changes in the result if you pick one over the other?
- Over what lifetime? Where will that number come from (source or assumption)?

**Functional unit:** Supply 1 kWh to one vehicle through a Level 2 AC wallbox for private or semi-public use, in accordance with the reference use scenario defined in PSR-0018 ed1.3 (S01, §3.1.1).

**Declared unit:** One Level 2 AC wallbox ensuring the recharging of electric vehicles over a reference service life of 10 years, in accordance with PSR-0018 ed1.3 (S01).

**Reference flow:** 1/28,251 of one Level 2 AC wallbox with a single charging point (≈ 3.54 × 10⁻⁵ units per kWh), based on 28,251 kWh supplied per charging point over the 10-year reference service life (S01, p. 20). The reference flow includes the charger, its attached charging cable (base case 24 ft / 7.3 m; see A02), its packaging, and the installation accessories supplied with the product (S01, §3.1.2).

**Derivation of 28,251 kWh** (S01, p. 20):

| Step | Value | Source |
|---|---|---|
| Average daily travel | 43 km/day | Enedis/BVA EV owner survey (2020), as cited in S01 |
| Share charged at private stations | 90% → 38.7 km/day | S01 |
| Distance over 10 years | 38.7 × 365 × 10 = 141,255 km | Calculated |
| Vehicle consumption | 20 kWh/100 km | S01 ("industrial data") |
| Energy supplied over 10 years | 141,255 × 20 / 100 = 28,251 kWh | Calculated |
